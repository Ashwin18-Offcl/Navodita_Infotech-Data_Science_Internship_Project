# Customer Review Sentiment Analysis using NLP
This notebook demonstrates sentiment classification using TF-IDF and Logistic Regression.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# Load dataset
df = pd.read_csv('reviews.csv')
df.head()
```

```python
# Features and Labels
X = df['review']
y = df['sentiment']
```

```python
# Convert text to numeric using TF-IDF
vectorizer = TfidfVectorizer()
X_vectorized = vectorizer.fit_transform(X)
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X_vectorized, y, test_size=0.2, random_state=42)
```

```python
# Train model
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
# Prediction
y_pred = model.predict(X_test)
print('Accuracy:', accuracy_score(y_test, y_pred))
```

```python
# Test with custom input
sample = ['This product is very good']
sample_vec = vectorizer.transform(sample)
prediction = model.predict(sample_vec)
print('Sentiment:', prediction[0])
```
